## Bizarrerie #1

Que donne cette addition ?

```js
"22" + 2
```

💡 **Coercion de type** : l'opérateur `+` voit une chaîne à gauche, donc il convertit `2` en chaîne et **concatène** au lieu d'additionner → `"222"`.

## Bizarrerie #2

Quel est le type de `NaN` ("Not a Number") ?

```js
typeof NaN
```

````javascript
typeof NaN
````

💡 **Un bug historique** : `NaN` est bien une valeur numérique spéciale (résultat d'une opération invalide comme `0/0`), donc `typeof NaN === "number"`. Autre piège lié : `NaN === NaN` vaut `false` — il faut utiliser `Number.isNaN(x)` pour le tester.

## Bizarrerie #3

Ce test d'égalité est-il vrai ?

```js
0.1 + 0.2 === 0.3
```

````javascript
0.1 + 0.2 === 0.3
````

💡 **Précision des flottants** : JavaScript stocke les nombres en virgule flottante binaire (IEEE 754), qui ne peut pas représenter exactement `0.1` ou `0.2`. Le résultat réel est `0.30000000000000004`. Pour comparer, on utilise une marge d'erreur (`Math.abs(a - b) < Number.EPSILON`) plutôt que `===`.
